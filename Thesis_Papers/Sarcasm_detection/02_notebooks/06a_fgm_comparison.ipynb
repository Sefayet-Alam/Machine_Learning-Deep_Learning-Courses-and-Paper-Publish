```python
from pathlib import Path
import pandas as pd
```

```python
TABLES = Path("../04_outputs/tables")

ml_df = pd.read_csv(TABLES / "baseline_ml_results.csv")
bert_df = pd.read_csv(TABLES / "banglabert_binary_summary.csv")
fgm_df = pd.read_csv(TABLES / "banglabert_fgm_ben_sarc_binary_results.csv")
```

```python
ml_ben = ml_df[ml_df["dataset"] == "ben_sarc_binary"].copy()
ml_row = {
    "model": "tfidf_logreg",
    "dataset": "ben_sarc_binary",
    "test_accuracy": float(ml_ben["test_accuracy"].iloc[0]),
    "test_macro_f1": float(ml_ben["test_macro_f1"].iloc[0]),
    "test_f1_binary": float(ml_ben["test_f1_binary"].iloc[0]),
}
```

```python
bert_ben = bert_df[
    (bert_df["dataset"] == "ben_sarc_binary") & (bert_df["split"] == "test")
].copy()

bert_row = {
    "model": "banglabert",
    "dataset": "ben_sarc_binary",
    "test_accuracy": float(bert_ben["accuracy"].iloc[0]),
    "test_macro_f1": float(bert_ben["macro_f1"].iloc[0]),
    "test_f1_binary": float(bert_ben["f1_binary"].iloc[0]),
}
```

```python
fgm_ben = fgm_df[fgm_df["split"] == "test"].copy()

fgm_row = {
    "model": "banglabert_fgm",
    "dataset": "ben_sarc_binary",
    "test_accuracy": float(fgm_ben["accuracy"].iloc[0]),
    "test_macro_f1": float(fgm_ben["macro_f1"].iloc[0]),
    "test_f1_binary": float(fgm_ben["f1_binary"].iloc[0]),
}
```

```python
comparison_df = pd.DataFrame([ml_row, bert_row, fgm_row])
comparison_df["macro_f1_gain_vs_tfidf"] = comparison_df["test_macro_f1"] - comparison_df.loc[
    comparison_df["model"] == "tfidf_logreg", "test_macro_f1"
].iloc[0]

comparison_df = comparison_df.sort_values("test_macro_f1", ascending=False).reset_index(drop=True)
comparison_df
```

```python
comparison_df.to_csv(TABLES / "ben_sarc_model_progression.csv", index=False)
print(TABLES / "ben_sarc_model_progression.csv")
```
