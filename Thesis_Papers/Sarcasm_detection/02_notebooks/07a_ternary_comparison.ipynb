```python
from pathlib import Path
import pandas as pd
```

```python
TABLES = Path("../04_outputs/tables")

plain_df = pd.read_csv(TABLES / "banglabert_banglasarc3_ternary_results.csv")
weighted_df = pd.read_csv(TABLES / "banglabert_weighted_banglasarc3_ternary_results.csv")
```

```python
plain_test = plain_df[plain_df["split"] == "test"].copy()
weighted_test = weighted_df[weighted_df["split"] == "test"].copy()

plain_row = {
    "model": "banglabert",
    "dataset": "banglasarc3_ternary",
    "test_accuracy": float(plain_test["accuracy"].iloc[0]),
    "test_macro_f1": float(plain_test["macro_f1"].iloc[0]),
    "test_weighted_f1": float(plain_test["weighted_f1"].iloc[0]),
}

weighted_row = {
    "model": "banglabert_weighted",
    "dataset": "banglasarc3_ternary",
    "test_accuracy": float(weighted_test["accuracy"].iloc[0]),
    "test_macro_f1": float(weighted_test["macro_f1"].iloc[0]),
    "test_weighted_f1": float(weighted_test["weighted_f1"].iloc[0]),
}
```

```python
comparison_df = pd.DataFrame([plain_row, weighted_row])
comparison_df["macro_f1_gain_vs_plain"] = comparison_df["test_macro_f1"] - comparison_df.loc[
    comparison_df["model"] == "banglabert", "test_macro_f1"
].iloc[0]

comparison_df = comparison_df.sort_values("test_macro_f1", ascending=False).reset_index(drop=True)
comparison_df
```

```python
comparison_df.to_csv(TABLES / "banglasarc3_ternary_model_comparison.csv", index=False)
print(TABLES / "banglasarc3_ternary_model_comparison.csv")
```
